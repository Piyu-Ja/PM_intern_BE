```python
import numpy as np
import pandas as pd
import scipy.sparse as sp
from sentence_transformers import SentenceTransformer
from implicit.als import AlternatingLeastSquares
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import MinMaxScaler
from pprint import pprint
import math
import time
```

```python
# ---------- PARAMETERS ----------
TOP_K = 10
ALS_FACTORS = 32
ALS_REG = 0.1
ALS_ITERS = 15
BASE_W_CONTENT = 0.4
BASE_W_DEMO = 0.2
BASE_W_CF = 0.4

def load_table(name):
    res = supabase.table(name).select("*").execute()
    if hasattr(res, "error") and res.error:
        raise RuntimeError(f"Error loading {name}: {res.error}")
    return pd.DataFrame(res.data)

def normalize_mat(m):
    if np.all(np.isclose(m, m.flat[0])):  # constant matrix
        return np.zeros_like(m)
    mn, mx = m.min(), m.max()
    if math.isclose(mx, mn):
        return np.zeros_like(m)
    return (m - mn) / (mx - mn)
```

```python
# ---------- 1) Load tables ----------
df_students = load_table("students")
df_skills = load_table("skills")
df_student_skills = load_table("student_skills")
df_internships = load_table("internships")
df_internship_skills = load_table("internship_skills")
# interactions may be empty/absent
try:
    df_interactions = load_table("interactions")
except Exception:
    df_interactions = pd.DataFrame(columns=["id","student_id","internship_id","event_type","timestamp"])

print("Loaded tables — rows:",
      "students", len(df_students),
      "internships", len(df_internships),
      "skills", len(df_skills),
      "student_skills", len(df_student_skills),
      "internship_skills", len(df_internship_skills),
      "interactions", len(df_interactions))
```

```python
# ---------- 2) Build skill lists for students & internships ----------
def norm_text(x):
    return None if pd.isnull(x) else str(x).strip().lower()

# maps: skill_id -> skill name (normalized)
skill_name_by_id = df_skills.set_index('id')['name'].apply(lambda s: norm_text(s)).to_dict() if not df_skills.empty else {}

# student_id -> list of skill names
if not df_student_skills.empty:
    merged = df_student_skills.merge(df_skills, left_on='skill_id', right_on='id', how='left')
    student_skill_map = merged.groupby('student_id')['name'].apply(lambda vals: [norm_text(v) for v in vals if pd.notnull(v)]).to_dict()
else:
    student_skill_map = {}

if not df_internship_skills.empty:
    merged2 = df_internship_skills.merge(df_skills, left_on='skill_id', right_on='id', how='left')
    intern_skill_map = merged2.groupby('internship_id')['name'].apply(lambda vals: [norm_text(v) for v in vals if pd.notnull(v)]).to_dict()
else:
    intern_skill_map = {}

# attach list columns (ensure consistent order of students/internships)
df_students['skills_list'] = df_students['id'].map(student_skill_map).apply(lambda x: x if isinstance(x, list) else [])
df_internships['skills_list'] = df_internships['id'].map(intern_skill_map).apply(lambda x: x if isinstance(x, list) else [])

```

```python
# ---------- 3) Content embeddings (skills + text) ----------
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
def embed_from_list(lst):
    if not lst:
        return np.zeros(embed_model.get_sentence_embedding_dimension(), dtype=float)
    text = ", ".join([str(x) for x in lst])
    return embed_model.encode(text)

# create embeddings for skill lists
df_students['skills_emb'] = df_students['skills_list'].apply(embed_from_list)
# optional: embed text columns (degree/major/experience)
df_students['text_concat'] = df_students[['degree','major','experience']].fillna('').astype(str).agg(' '.join, axis=1)
df_students['text_emb'] = df_students['text_concat'].apply(lambda s: embed_model.encode(s if s.strip()!="" else ""))
df_students['content_emb'] = df_students.apply(lambda r: np.concatenate([r['skills_emb'], r['text_emb']]), axis=1)

df_internships['skills_emb'] = df_internships['skills_list'].apply(embed_from_list)
df_internships['text_concat'] = df_internships[['title','company','location','duration']].fillna('').astype(str).agg(' '.join, axis=1)
df_internships['text_emb'] = df_internships['text_concat'].apply(lambda s: embed_model.encode(s if s.strip()!="" else ""))
df_internships['content_emb'] = df_internships.apply(lambda r: np.concatenate([r['skills_emb'], r['text_emb']]), axis=1)

print("Embeddings ready — dim:", df_students['content_emb'].iloc[0].shape if len(df_students)>0 else "0")
```

```python
# ---------- 4) Demographic features (one-hot small set) ----------
# choose columns available: location, major, degree (if present)
demo_s_cols = []
demo_i_cols = []
if 'location' in df_students.columns:
    demo_s_cols.append('location')
if 'major' in df_students.columns:
    demo_s_cols.append('major')
if 'degree' in df_students.columns:
    demo_s_cols.append('degree')

if 'location' in df_internships.columns:
    demo_i_cols.append('location')
if 'duration' in df_internships.columns:
    demo_i_cols.append('duration')

# create dummy frames and align columns
if demo_s_cols:
    s_demo = pd.get_dummies(df_students[['id'] + demo_s_cols].fillna(''), columns=demo_s_cols).set_index('id')
else:
    s_demo = pd.DataFrame(index=df_students['id'])

if demo_i_cols:
    i_demo = pd.get_dummies(df_internships[['id'] + demo_i_cols].fillna(''), columns=demo_i_cols).set_index('id')
else:
    i_demo = pd.DataFrame(index=df_internships['id'])

# intersect columns to compute similarity (safe)
common_demo_cols = list(set(s_demo.columns).intersection(set(i_demo.columns)))
if len(common_demo_cols) == 0:
    X_demo_students = np.zeros((len(df_students), 1))
    X_demo_internships = np.zeros((len(df_internships), 1))
else:
    X_demo_students = s_demo.reindex(df_students['id'])[common_demo_cols].fillna(0).values
    X_demo_internships = i_demo.reindex(df_internships['id'])[common_demo_cols].fillna(0).values

print("Demographic matrices shape:", X_demo_students.shape, X_demo_internships.shape)
```

```python
# ---------- 5) Build interaction sparse matrix for CF ----------
# map to indices
student_ids = df_students['id'].tolist()
intern_ids = df_internships['id'].tolist()
student_index = {sid: idx for idx, sid in enumerate(student_ids)}
intern_index = {iid: idx for idx, iid in enumerate(intern_ids)}

# map event types to numeric weights
event_map = {'click': 1.0, 'save': 2.0, 'apply': 3.0}
if df_interactions.empty:
    interaction_sparse = sp.csr_matrix((len(student_ids), len(intern_ids)))
else:
    # keep only interactions that map to known users/items
    df_inter_temp = df_interactions.copy()
    df_inter_temp = df_inter_temp[df_inter_temp['student_id'].isin(student_index) & df_inter_temp['internship_id'].isin(intern_index)]
    if df_inter_temp.empty:
        interaction_sparse = sp.csr_matrix((len(student_ids), len(intern_ids)))
    else:
        rows = df_inter_temp['student_id'].map(student_index).astype(int)
        cols = df_inter_temp['internship_id'].map(intern_index).astype(int)
        weights = df_inter_temp['event_type'].map(lambda e: event_map.get(e, 1.0)).astype(float)
        interaction_sparse = sp.csr_matrix((weights, (rows, cols)), shape=(len(student_ids), len(intern_ids)))

print("Interaction sparse shape:", interaction_sparse.shape, "nnz:", interaction_sparse.nnz)
```

```python
# --------- 6) Train ALS (collaborative) ----------
if interaction_sparse.nnz > 0:
    als = AlternatingLeastSquares(factors=ALS_FACTORS, regularization=ALS_REG, iterations=ALS_ITERS)
    als.fit(interaction_sparse.T)  # implicit expects items x users
    user_factors = als.user_factors  # shape: n_users x factors
    item_factors = als.item_factors  # shape: n_items x factors
else:
    user_factors = np.zeros((len(student_ids), ALS_FACTORS))
    item_factors = np.zeros((len(intern_ids), ALS_FACTORS))
```

```python
# ---------- 7) Compute similarity matrices ----------
if len(df_students) == 0 or len(df_internships) == 0:
    raise RuntimeError("No students or internships present.")

content_mat = cosine_similarity(np.vstack(df_students['content_emb'].values),
                                np.vstack(df_internships['content_emb'].values))

demo_mat = cosine_similarity(X_demo_students, X_demo_internships) if X_demo_students.shape[1] > 0 else np.zeros_like(content_mat)

cf_mat = np.dot(user_factors, item_factors.T) if user_factors.size and item_factors.size else np.zeros_like(content_mat)

# normalize
C = normalize_mat(content_mat)
D = normalize_mat(demo_mat)
F = normalize_mat(cf_mat)
```

```python
# ---------- 8) Hybrid scoring with per-user dynamic weighting ----------
hybrid = np.zeros_like(C)
for i in range(C.shape[0]):
    # history count
    history_count = int((interaction_sparse[i].nnz) if interaction_sparse.shape[1]>0 else 0)
    if history_count == 0:
        w_cf = 0.0
        # redistribute CF weight equally to content and demo (you can tune)
        w_c = BASE_W_CONTENT + 0.5 * BASE_W_CF
        w_d = BASE_W_DEMO + 0.5 * BASE_W_CF
    else:
        # scale CF weight based on history (simple heuristic)
        cf_ratio = min(history_count / 10.0, 1.0)
        w_cf = BASE_W_CF * cf_ratio
        remaining = 1.0 - w_cf
        # keep relative proportion of content vs demo same as base
        total_base = BASE_W_CONTENT + BASE_W_DEMO
        w_c = remaining * (BASE_W_CONTENT / total_base)
        w_d = remaining * (BASE_W_DEMO / total_base)
    hybrid[i, :] = w_c * C[i, :] + w_d * D[i, :] + w_cf * F[i, :]

print("Hybrid matrix computed. Shape:", hybrid.shape)
```

```python
# ---------- 9) Prepare and view Top 5 Hybrid Recommendations Overall ----------

TOP_K_DISPLAY = 5  # Show only top 5 recommendations overall

rows_to_upsert = []
all_recs = []  # store all for sorting

for i, sid in enumerate(student_ids):
    for j, score_value in enumerate(hybrid[i, :]):
        rec = {
            "student_id": sid,
            "internship_id": intern_ids[j],
            "score": float(score_value)
        }
        rows_to_upsert.append(rec)
        all_recs.append(rec)

# Upsert all recommendations into Supabase
res = supabase.table("recommendations").upsert(rows_to_upsert).execute()
print(" Upsert complete! Status Code:", getattr(res, 'status_code', None))

# ---------- Sort and Display Top 5 Overall ----------

# Sort all recommendations by score descending
all_recs_sorted = sorted(all_recs, key=lambda x: x['score'], reverse=True)

# Take top 5
top5_recs = all_recs_sorted[:TOP_K_DISPLAY]

print("\n Top 5 Internship Recommendations Overall (by Hybrid Fit Score):\n")
for rank, rec in enumerate(top5_recs, 1):
    print(
        f"Rank: {rank} | "
        f"Student: {rec['student_id']} | "
        f"Internship: {rec['internship_id']} | "
        f"Fit Score: {round(rec['score'], 3)}"
    )

print("\n Display complete! These are the top 5 hybrid-based recommendations overall.")
```
